```python
import pandas as pd
import matplotlib.pyplot as plt

history_df = pd.read_csv("history.csv")
pareto_df = pd.read_csv("pareto_front.csv")

history_df.head()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(history_df["generation"], history_df["best_objective_1"], label="Best Obj 1")
plt.plot(history_df["generation"], history_df["best_objective_2"], label="Best Obj 2")
plt.xlabel("Generation")
plt.ylabel("Best Objective Value")
plt.title("Best Objectives Over Generations")
plt.legend()
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(history_df["generation"], history_df["avg_objective_1"], label="Avg Obj 1")
plt.plot(history_df["generation"], history_df["avg_objective_2"], label="Avg Obj 2")
plt.xlabel("Generation")
plt.ylabel("Average Objective Value")
plt.title("Average Objectives Over Generations")
plt.legend()
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(history_df["generation"], history_df["avg_penalties"])
plt.xlabel("Generation")
plt.ylabel("Average Penalties")
plt.title("Penalty Convergence")
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(pareto_df["objective_1"], pareto_df["objective_2"])

plt.xlabel("Objective 1")
plt.ylabel("Objective 2")
plt.title("Final Pareto Front")
plt.grid()
plt.show()
```
